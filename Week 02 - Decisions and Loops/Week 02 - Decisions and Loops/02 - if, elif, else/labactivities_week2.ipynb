```python
fruit = ['apple', 'banana', 'cherry', 'durian', 'elderberry']
print(fruit[1])
print(fruit[-2])
print(fruit[:2:1])
print(fruit[3::])
print(fruit[1:4:1])
(fruit.append("strawberry")) #adds another element to the list
print(fruit)
fruit.remove("durian") #removes an element from the list
print(fruit)
print(fruit.pop())
```
